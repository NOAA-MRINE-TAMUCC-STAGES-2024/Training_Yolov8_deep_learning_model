```python
import os
import shutil

# Define the paths
train_folder = r"C:\MS_2024\Marine Stages\Airial images from NOAA\Yolo training satalight quality\yolo training satalight quality\images\train"
val_folder = r"C:\MS_2024\Marine Stages\Airial images from NOAA\Yolo training satalight quality\yolo training satalight quality\images\Val"

# Get the list of filenames in both folders
train_files = set(os.listdir(train_folder))
val_files = set(os.listdir(val_folder))

# Find the common files
common_files = train_files.intersection(val_files)

# Delete the common files from the train folder
for file_name in common_files:
    file_path = os.path.join(train_folder, file_name)
    os.remove(file_path)
    print(f"Deleted {file_name} from train folder.")
```

```python
import os
import shutil

# Define paths
train_labels_path = r"C:\MS_2024\Marine Stages\Airial images from NOAA\Yolo training satalight quality\yolo training satalight quality\labels\train"
val_labels_path = r"C:\MS_2024\Marine Stages\Airial images from NOAA\Yolo training satalight quality\yolo training satalight quality\labels\Val"

# List of image filenames that were moved
moved_images = [
    "2022-01-12_B2213168_mara_Alt600.JPG",
    "Copy of MR-10Apr12-002 color adjust.JPG",
    "2022-02-12_A2212877_mara_Alt600.JPG",
    "2022_01_12_A2204359_mara_Alt600.JPG",
    "Copy of MR-17Mar12-086.JPG",
    "2022-01-24_A2201102_mara_hhsh_Alt1500.JPG"
]

# Move corresponding label files
for image in moved_images:
    label_file = os.path.splitext(image)[0] + ".txt"  # Replace image extension with .txt
    source_label_path = os.path.join(train_labels_path, label_file)
    destination_label_path = os.path.join(val_labels_path, label_file)
    
    # Check if the label file exists and move it
    if os.path.exists(source_label_path):
        shutil.move(source_label_path, destination_label_path)
        print(f"Moved {label_file} from train to val.")
    else:
        print(f"Label file {label_file} not found in train folder.")
```
